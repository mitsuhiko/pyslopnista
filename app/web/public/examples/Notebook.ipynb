# A notebook

A notebook mixes notes and code in **cells**. Tap ▶ on a cell to run it, or press Shift-Return on a keyboard. Run All (at the top) runs them in order.

What a cell defines stays around for the cells after it: they share one Python, the *kernel*. Restart gives you a fresh one.

```python
import math

radius = 3
area = math.pi * radius**2
area
```

The value of a cell's last line shows below it (end the line with `;` to hide it). `print()` works too:

```python
for r in range(1, 4):
    print(f"r={r}: area {math.pi * r**2:.2f}")
```

## Plots

Figures show up below the cell, no `plt.show()` needed.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 4 * np.pi, 200)
plt.plot(x, np.sin(x), label="sin")
plt.plot(x, np.cos(x), label="cos")
plt.legend()
plt.title("Waves");
```

## Tables and rich output

pandas tables, HTML and Markdown show as they would in Jupyter. `display()` shows more than one thing from a cell.

```python
import pandas as pd

df = pd.DataFrame({"planet": ["Mercury", "Venus", "Earth", "Mars"], "moons": [0, 0, 1, 2], "day (h)": [4222.6, 2802.0, 24.0, 24.7]})
df
```

```python
from IPython.display import HTML, Markdown

display(Markdown(f"The planets have **{df.moons.sum()}** moons together."))
display(HTML("<p style='color: tomato'>This line is <b>HTML</b>.</p>"))
```
